```python
# Teranry Operator
# syntax
# express_1 if<condition> else express_2
```

```python
gender=['Male','Female','Female','Male']
# Convert male to 0 & female to 1
[0 if val=='Male' else 1 for val in gender]
```

```python
salary=[24000,12500,45000,65000,11000,10000]
# PS: if sal>25000 then 10% bonus else 20%
[sal+sal*0.1 if sal>25000 else sal+sal*.2 for sal in salary]
```

```python
#  Remove Vowels from a String
text = "interview preparation"
vowels = "aeiouAEIOU"
result=[ch for ch in text if ch not in vowels]
print(result)
```

```python
text = "interview preparation"
vowels = "aeiouAEIOU"
result = "".join([ch for ch in text if ch not in vowels])
print(result)
```

```python
# Find Common Elements Between Two Lists
a = [1, 2, 3, 4, 5]
b = [3, 4, 5, 6, 7]
comman=[x for x in a if x in b]
print(comman)
```

```python
#  Flatten a 2D List
data = [[10, 20], [30, 40], [50, 60]]
flat = [item for row in data for item in row]
print(flat)
```

```python
data = [[10, 20], [30, 40], [50, 60]]
flat = [item for column in data for item in column]
print(flat)
```

```python
#  Find Words Starting with a Specific Letter P
words = ["Python", "Java", "Pandas", "SQL", "PowerBI"]
letter = [word for word in words if word.startswith('P')]
print(letter)
```
