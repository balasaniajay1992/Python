```python
# 3. Transfer statement
# 1. break -exit from current loop
# 2. Continue - skip current condition and continuw the loop
# 3. pass - used to create an empty blocks
```

```python
f=['apple','apple','cherry','grapes','mango','banana']
for item in f:
  if item=='grapes':
    break
  else:
    print('Process:',item)
```

```python
f=['apple','apple','cherry','grapes','mango','banana']
for item in f:
  if item !='grapes':
    continue
  else:
    print('Process:',item)
```

```python
# Continue: will help to continue the loop and skip current condition or item
products = ['mobile','laptop','headphone','mouse','keyboard','mobile','laptop','headphone']

search_item = input('Enter item to search:')

for item in products:
  print('Checking:',item)

  if item == search_item:
    print('Product found:',item)
    continue
```

```python
# List Comprehension
# to create a new list in simple onliner solution
# it uses the treneary opertaor to check if else option
```

```python
# syntax:[operation for var in sequence/iteration]
d=[2,3,4,5,6,7]
# create a new list with squared numbers
[num*num for num in d]
```

```python
nms=['prased','imran','gayatri','jay','bhavana']
[nm.title() for nm in nms]
```

```python
nms=['prased','imran','gayatri','jay','bhavana']
[(nm.title(),len(nm)) for nm in nms]
```

```python
# Conditional list comprehension
# fetech whos length  len is >6
# syntax:[operation for var in sequence/iteration if<condition>]
# fisrt element  fetched in nm...> check condition len(nm)>6...> perform opertaion
[nm.title() for nm in nms if len(nm)>6]
```
