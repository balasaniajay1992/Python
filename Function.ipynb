```python
# 4 Variable length args
# 2types
# 1. Variable length postional args
# 2. Variable length keyword args
```

```python
# 1. Variable length postional args
def test(n):
  print(n)

test(10)
```

```python
test(10,20,30)
```

```python
# when user doesn't have a strict requirments of sending the values to the arguments
#We need to prefix the arguments with *
def test(*n): # *args
  print(n)

test(10)
```

```python
test(1,2,3,4,)
```

```python
# 2. Variable length keyword args--> **kwargs
def test(**x):
  print(x)

test(n1=1,n2=2,n3=3,n4=4)
```

```python
test(name='ajay',age=32,sal=40000)
```

```python
test(n1=1,n4=2,n2=3,n3=4)
```

```python
# When we awant to combined both we can do so but remember the rule
# fisrt we need postional n then keyword
def sample(*args,**kwargs):
  print(args,kwargs)

sample(10,20,30,30,num1=200,num2=400)
```
