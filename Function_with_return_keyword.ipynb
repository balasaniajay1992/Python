```python
# Positional, Keyword, Default and Var.Len.arguments
# Function with return keyword
# Function will work on specific requirments
# it will generate one results
# this result we need to use as an input to the outside of the function
# for this purpose we need to use return keyword

def test(balance):
  if balance <= 0:
    return 'No'
  else:
    return 'Yes'

test(120)
```

```python
# Purchase and billing example using return
items = {'rice':56,'chana':30,'wheat':60,'parle_g':10,'milk':35}
bill = []
def purchase(*args):
  for item in args:
    #print(item,items[item])
    bill.append(items[item])
  return bill

total = purchase('wheat')
print('Item amounts:',total)
```

```python
# purchase and billing example using return
items = {'rice':56,'chana':30,'wheat':60,'parle_g':10,'milk':35}
bill = []
def purchase(*args):
  for item in args:
    #print(item,items[item])
    bill.append(items[item])
  return bill

def check(total):
  return sum(total)

# calling of purchase
total = purchase('wheat','milk')
print('Item amounts:',total)

# calling of check
ans = check(total)
print('Final bill amoutnt is Rs:',ans)
```
