```python
# Lambda Function is called as Anonymous/nameless function
# it is one liner function
# Only expressiom is allowed in lambda
# syntax: lambda parameter/s:expression
```

```python
# squaring of number
sq=lambda num:num*num
sq(8)
```

```python
# Lambda function by default returns the value of expression
change = lambda name:name.upper()

result=change('ajay')
print(result)
```

```python
# Higher order function
# It is is a function who need another function as an input
# function(function)
```

```python
# map, filter and reduce are high order function
```

```python
# iterable - collection of elements --> set,list,tuple,set,dict,range,array ...
# map(function,iterable)
k = [20,5,4,60,10]

# apply an operation over each element from an iterable

map(lambda num: num*num,k)
```

```python
# Rule - when we are getting output in the form of object the we have 2 choices
# Choice1- typecast the object to see actual value
tuple(map(lambda num: num*num,k))
```

```python
k
```

```python
# Choice2 - use for loop
out = map(lambda num:num*num,k)
for val in out:
  print(val)
```

```python
# can we use more arguments in lambda: Yes
test=lambda a,b,c:a+b
test(100,800,500)
```

```python
test=lambda a,b,c:max(a,b,c)
test(100,800,500)
```

```python
names=('gayatri','abdul','snehal','jay')
# conver the above name in upper caseusing map()
out = list(map(lambda nm:nm.upper(),names))
print(out)
```

```python
upper_names = tuple(map(lambda x: x.upper(), names))

print(upper_names)
```

```python
list(map(lambda name: name.upper(), names))
```

```python
# Assignment I need to count occurance of each word in str
# s='simple dimple'
s = "simple dimple"

result = dict(map(lambda ch: (ch, s.count(ch)), set(s)))

print(result)
```

```python
# Filter(function,iterable)
# this is condition function
# the value for ehich condition is Ture will get fetched and operation will be perfromed if requires
p = (12,-33,40,-8,99,-12)
# filter -ve numbers
filter(lambda num:num<0,p)
```

```python
list(filter(lambda num:num<0,p)) # -ve number
```

```python
list(filter(lambda num:num>0,p)) # +ve number
```
