```python
# 2. Iterative statements
# It contains 2 loops
#1. for loop
#2. while loop
```

```python
# for loop: is an iterative loop which fetches one element at a time from the collection inside the block
a=[10,20,30,40]

# syntax for var in collection/iterable
for val in a:
  print(val)
```

```python
# Take a dict and apply for loop
a={'fruit':'Mango',
   'Veg':'Tamato',
   'Drinks':'Thumsup'
   }

for key,value in a.items():
  print(key,value)
```

```python
f={1:100,2:200}
for val in f:
  print(f[val])
```
